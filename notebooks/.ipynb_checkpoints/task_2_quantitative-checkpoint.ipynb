Seaborn Correlation Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
sns.heatmap(data[['Close', 'SMA_20', 'SMA_50', 'RSI', 'MACD', 'MACD_Signal', 'Daily_Return']]
            .corr(),
            annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Indicators")
plt.show()
```

Cumulative Return Curve

```python
plt.figure(figsize=(12,5))
plt.plot(data.index, data["Cumulative_Return"], label="Cumulative Return")
plt.title(f"{ticker} – Cumulative Return")
plt.legend()
plt.show()
```
